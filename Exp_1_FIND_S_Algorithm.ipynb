```python
import pandas as pd

# Create the CSV file inside Colab
csv_text = """Sky,AirTemp,Humidity,Wind,Water,Forecast,EnjoySport
Sunny,Warm,Normal,Strong,Warm,Same,Yes
Sunny,Warm,High,Strong,Warm,Same,Yes
Rainy,Cold,High,Strong,Warm,Change,No
Sunny,Warm,High,Strong,Cool,Change,Yes
"""
with open('enjoysport.csv', 'w') as f:
    f.write(csv_text)

data = pd.read_csv('enjoysport.csv')
print("Training data:\n", data, "\n")

X = data.iloc[:, :-1].values
y = data.iloc[:, -1].values

hypothesis = None
for i, (x, label) in enumerate(zip(X, y), 1):
    if label == 'Yes':                      # FIND-S ignores negative examples
        if hypothesis is None:
            hypothesis = list(x)            # start with first positive example
        else:
            hypothesis = [h if h == xi else '?' for h, xi in zip(hypothesis, x)]
    print(f"After example {i}: {hypothesis}")

print("\nMost specific hypothesis:", hypothesis)
```
